# Example: exploring the sales pipeline

Notebooks are for **exploring**: looking at data, trying ideas and making charts.
Code you want to keep and reuse belongs in `src/ds_starter/`, and you import it here.

Run each cell with **Shift + Enter**.

```python
import pandas as pd

from ds_starter.config import RAW_SALES_PATH
from ds_starter.helpers import add_revenue, clean_sales
from ds_starter.pipeline import run_pipeline
```

## 1. Read the raw data

```python
raw_sales = pd.read_csv(RAW_SALES_PATH)
raw_sales.head()
```

```python
raw_sales.info()
```

## 2. Try one helper at a time

The raw data is a little messy: some regions are in lower case, some rows are missing `units`, and two orders are duplicated. `clean_sales` fixes that.

```python
raw_sales["region"].value_counts()
```

```python
cleaned = clean_sales(raw_sales)
cleaned["region"].value_counts()
```

```python
with_revenue = add_revenue(cleaned)
with_revenue.head()
```

## 3. Run the whole pipeline

`run_pipeline` is the orchestrator. It calls every helper in order and returns the summary table.

```python
summary = run_pipeline(raw_sales)
summary
```

Try answering a new question with the cleaned data, for example: which product earns the most revenue?
